# Loan Risk Assessment System - Data Analysis
This notebook explores `train.csv` and reproduces the key steps used in `train_model.py`.
Run this from the project root, or adjust the path below.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/train.csv')
df.head()
```

```python
print('Shape:', df.shape)
print(df.dtypes)
print('Missing values:\n', df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
print(df['Risk_Flag'].value_counts())
sns.countplot(x='Risk_Flag', data=df)
plt.title('Target Class Distribution')
plt.show()
```

```python
numeric_cols = ['Income', 'Age', 'Experience', 'CURRENT_JOB_YRS', 'CURRENT_HOUSE_YRS']
df[numeric_cols].hist(figsize=(10, 8), bins=30)
plt.tight_layout()
plt.show()
```

```python
sns.heatmap(df[numeric_cols + ['Risk_Flag']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
